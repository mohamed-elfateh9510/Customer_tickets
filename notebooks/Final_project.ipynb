**تنظيف النصوص واستخراج الخصائص وتدريب النماذج**

في هذه المرحلة يتم تجهيز نصوص الـ Tickets لتكون مناسبة للاستخدام في نماذج التعلم الآلي. يتم تنظيف النص من خلال تحويل جميع الحروف إلى أحرف صغيرة، وإزالة الرموز والأرقام، وإزالة كلمات التوقف الإنجليزية (Stopwords)، واستبعاد الكلمات القصيرة جدًا.

تم تدريب الـ TF-IDF على بيانات التدريب فقط باستخدام fit_transform، ثم تم تحويل بيانات الاختبار باستخدام transform فقط. وهذا يساعد على منع حدوث Data Leakage.

بعد استخراج الخصائص، تم تدريب نموذجين باستخدام
LinearSVC:

النموذج الأول للتنبؤ بـ Ticket Type.
النموذج الثاني للتنبؤ بـ Ticket Priority.

وأخيرًا تم حفظ الـ TF-IDF Vectorizer والنموذجين باستخدام joblib، بحيث يمكن إعادة استخدامهم لاحقًا لإجراء Predictions جديدة دون الحاجة إلى إعادة التدريب.

````python
import re
import nltk
from sklearn.model_selection import train_test_split
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib
from sklearn.svm import LinearSVC

nltk.download('stopwords', quiet =True)
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()
    words = [w for w in words if w not in stop_words and len(w) > 2]
    return " ".join(words)

df_clean['proccesed_text'] = df_clean['full_text'].apply(clean_text)

X = df_clean['proccesed_text']
y_cat = df_clean['ticket_type']
y_prio = df_clean['ticket_priority'] 


X_train, X_test, y_cat_train, y_cat_test = train_test_split(X, y_cat, random_state=24, test_size=0.2, stratify=y_cat)
_, _, y_prio_train, y_prio_test = train_test_split(X, y_prio, random_state=24, test_size=0.2, stratify=y_prio)


tf_idf = TfidfVectorizer(max_features=3000, ngram_range=(1, 2))

X_train_tfidf = tf_idf.fit_transform(X_train) 
X_test_tfidf = tf_idf.transform(X_test)       


cat_model = LinearSVC(random_state=24)
cat_model.fit(X_train_tfidf, y_cat_train)


prio_model = LinearSVC(random_state=24)
prio_model.fit(X_train_tfidf, y_prio_train)


joblib.dump(tf_idf, 'tfidf_vectorizer.pkl')
joblib.dump(cat_model, 'category_model.pkl')
joblib.dump(prio_model, 'priority_model.pkl')

print("Done! All models trained on the exact same TF-IDF features.")

````

**تدريب وتقييم نموذج Logistic Regression**

في هذه المرحلة يتم استخدام نموذج Logistic Regression كنموذج تصنيف إضافي للتنبؤ بكل من نوع الـ Ticket وأولوية الـ Ticket.

تم تدريب نموذجين منفصلين باستخدام نفس الـ TF-IDF Features التي تم استخراجها في المرحلة السابقة:

النموذج الأول للتنبؤ بـ ticket_type.
النموذج الثاني للتنبؤ بـ ticket_priority.

تم استخدام max_iter=1000 لإعطاء خوارزمية التدريب عددًا كافيًا من التكرارات للوصول إلى التقارب، بينما يساعد random_state=42 على الحصول على نتائج قابلة لإعادة الإنتاج.

بعد تدريب النماذج، يتم استخدام predict() لإجراء التنبؤات على بيانات الاختبار. ثم يتم تقييم أداء النماذج باستخدام Accuracy و Classification Report، والذي يعرض مقاييس Precision وRecall وF1-score وSupport لكل فئة.

في النهاية يتم حفظ نماذج Logistic Regression، بالإضافة إلى الـ TF-IDF Vectorizer وبعض بيانات التدريب، باستخدام joblib. ويسمح ذلك بإعادة تحميل هذه الملفات لاحقًا داخل التطبيق (app2.py) واستخدام النماذج للتنبؤ ببيانات Tickets جديدة دون الحاجة إلى إعادة تدريبها.

````python
import joblib
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report


log_reg_cat = LogisticRegression(max_iter=1000, random_state=42)
log_reg_cat.fit(X_train_tfidf, y_cat_train)

y_pred_cat = log_reg_cat.predict(X_test_tfidf)

print('=== Category Model Evaluation ===')
print('Accuracy of Logistic Regression (Category):', accuracy_score(y_cat_test, y_pred_cat))
print('-------------------------------------------')
print('Classification Report (Category):')
print(classification_report(y_cat_test, y_pred_cat))


log_reg_prio = LogisticRegression(max_iter=1000, random_state=42)
log_reg_prio.fit(X_train_tfidf, y_prio_train)

y_pred_prio = log_reg_prio.predict(X_test_tfidf)

print('\n=== Priority Model Evaluation ===')
print('Accuracy of Logistic Regression (Priority):', accuracy_score(y_prio_test, y_pred_prio))
print('-------------------------------------------')
print('Classification Report (Priority):')
print(classification_report(y_prio_test, y_pred_prio))


joblib.dump(log_reg_cat, 'category_model.pkl')
joblib.dump(log_reg_prio, 'priority_model.pkl')
joblib.dump(tf_idf, 'tfidf_vectorizer.pkl')
joblib.dump(X_train_tfidf, 'X_train_tf_idf.pkl')
joblib.dump(X_train, 'X_train.pkl')
joblib.dump(y_cat_train, 'y_train.pkl')

print("\nAll models and artifacts saved successfully for app2.py!")
````

**البحث عن التذاكر المشابهة باستخدام Cosine Similarity**

في هذه المرحلة يتم إنشاء نظام للبحث عن **أكثر الـ Tickets تشابهًا** مع Ticket جديدة، وذلك بالاعتماد على تمثيل النصوص باستخدام **TF-IDF**.

تستقبل الدالة `find_similar_tickets()` رسالة جديدة، ثم تطبق عليها نفس عملية تنظيف النصوص المستخدمة أثناء تدريب النماذج. بعد ذلك يتم تحويل الرسالة المنظفة إلى متجه TF-IDF باستخدام الـ TF-IDF Vectorizer الذي تم تدريبه مسبقًا.

بعد ذلك يتم حساب **Cosine Similarity** بين متجه الـ Ticket الجديدة وجميع متجهات الـ Tickets الموجودة في بيانات التدريب. وتوضح قيمة الـ Similarity مدى تشابه كل Ticket موجودة مع الرسالة الجديدة.

ثم يتم ترتيب نتائج التشابه واختيار أعلى `N` Tickets تشابهًا. ولكل Ticket مشابه يتم عرض النص، ونوع الـ Ticket، والأولوية، بالإضافة إلى **Similarity Score**.

في النهاية يتم استخدام رسالة تجريبية لتوضيح كيفية قيام النظام بالبحث عن الـ Tickets التاريخية الأكثر تشابهًا مع الطلب الجديد.

يمكن استخدام هذا الأسلوب في **البحث عن الحالات السابقة المشابهة، واقتراح الـ Tickets ذات الصلة، ومساعدة موظفي خدمة العملاء** من خلال عرض الحالات التي تم التعامل معها سابقًا وتشبه الطلب الحالي.


````python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

def find_similar_tickets(new_message, top_n=3):
    cleaned_msg = clean_text(new_message)
    
    message_vectorized = tf_idf.transform([cleaned_msg])
    
    similarities = cosine_similarity(message_vectorized, X_train_tfidf).flatten()
    top_indices = similarities.argsort()[-top_n:][::-1]
    
    results = []
    for idx in top_indices:
        results.append({
            'Ticket Text': X_train.iloc[idx],
            'Category': y_cat_train.iloc[idx],     
            'Priority': y_prio_train.iloc[idx],     
            'Similarity Score': round(similarities[idx], 4)
        })
    return pd.DataFrame(results)

sample_message = "I was charged twice on my credit card for the same order."
find_similar_tickets(sample_message)
````
