```python
# Data Cleaning
df_clean = df.copy()
df_clean['full_text'] = df_clean['Ticket Subject'].astype(str) + ' : ' + df_clean['Ticket Description'].astype(str)
print(df_clean['full_text'])
print('The count of duplicates in full_text coulmn : ', df_clean['full_text'].duplicated().sum())
```

```python
df_clean
```

```python
df_clean = df_clean.drop_duplicates(subset = 'full_text')
print('The count of duplicates in full_text coulmn : ', df_clean['full_text'].duplicated().sum())
selected_cols = ['full_text', 'Ticket Type', 'Ticket Priority', 'Ticket Status']
df_clean = df_clean[selected_cols].reset_index(drop =True)
df_clean
```

```python
import re
import nltk
from sklearn.model_selection import train_test_split
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer

nltk.download('stopwords', quiet =True)
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()
    words = [w for w in words if w not in stop_words and len(w) > 2]
    return " ".join(words)

df_clean['proccesed_text'] = df_clean['full_text'].apply(clean_text)
X = df_clean['proccesed_text']
y = df_clean['Ticket Type']

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=24, test_size =0.2, stratify=y)

tf_idf = TfidfVectorizer(max_features = 3000, ngram_range=(1, 2))

X_train_tf_idf = tf_idf.fit_transform(X_train)
X_test_tf_idf = tf_idf.transform(X_test)

print('Training tf_idf data shape is : ', X_train_tf_idf.shape)
print('Testing  tf_idf data shape is : ', X_test_tf_idf.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

log_reg = LogisticRegression(max_iter = 1000, random_state = 42)
log_reg.fit(X_train_tf_idf, y_train)

y_pred_log_reg = log_reg.predict(X_test_tf_idf)
print('accuracy of logistic regression :', accuracy_score(y_test, y_pred_log_reg))
print('---------------------------')
print('classification report :')
print(classification_report(y_test, y_pred_log_reg))
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

nb_model = MultinomialNB()
nb_model.fit(X_train_tf_idf, y_train)
y_pred_nb = nb_model.predict(X_test_tf_idf)
print(f'Accuracy naive bayes: {accuracy_score(y_test, y_pred_nb):.3f}')


svm_model = LinearSVC(random_state = 42)
svm_model.fit(X_train_tf_idf, y_train)
y_pred_svm = svm_model.predict(X_test_tf_idf)
print(f'Accuracy svm: {accuracy_score(y_test, y_pred_svm):.3f}')
print('accuracy of logistic regression :', accuracy_score(y_test, y_pred_log_reg))
```

```python
from sklearn.metrics.pairwise import cosine_similarity
def find_similar_tickets(new_message, top_n=3):
    cleaned_msg= clean_text(new_message)
    massege_vectorized = tf_idf.transform([cleaned_msg])
    similarities = cosine_similarity(massege_vectorized, X_train_tf_idf).flatten()
    top_indices = similarities.argsort()[-top_n:][::-1]
    results =[]
    for idx in top_indices:
        results.append({
            'Ticket Text': X_train.iloc[idx],
            'Category': y_train.iloc[idx],
            'Similarity Score': round(similarities[idx], 4)
        })
    return pd.DataFrame(results)

sample_message = "I was charged twice on my credit card for the same order."
find_similar_tickets(sample_message)
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report, accuracy_score

X_all = df_clean['proccesed_text']
y_priority = df_clean['Ticket Priority']

X_train_p, X_test_p, y_train_p, y_test_p = train_test_split(X_all, y_priority, test_size=0.2
                                                            , random_state=42, stratify= y_priority)
X_train_p_tfidf = tf_idf.fit_transform(X_train_p)
X_test_p_tfidf = tf_idf.transform(X_test_p)

priority_model = LinearSVC(random_state=42)
priority_model.fit(X_train_p_tfidf, y_train_p)

y_pred_p = priority_model.predict(X_test_p_tfidf)
print(f"Priority Model Accuracy: {accuracy_score(y_test_p, y_pred_p):.2f}")
print(classification_report(y_test_p, y_pred_p))
```

```python
def analyze_new_ticket(ticket_text):
    cleaned_msg = clean_text(ticket_text)
    mes_vec = tf_idf.transform([cleaned_msg])

    pred_category = svm_model.predict(mes_vec)[0]
    pred_priority = priority_model.predict(mes_vec)[0]

    similarities = cosine_similarity(mes_vec, X_train_tf_idf).flatten()
    top_indices = similarities.argsort()[-3:][::-1]

    print("=" * 50)
    print(f" New Ticket: '{ticket_text}' ")
    print(f" Predicted Category: {pred_category}")
    print(f" Predicted Priority: {pred_priority}") 
    print("=" * 50)
    print(" Top 3 Similar Past Tickets:")

    for i, idx in enumerate(top_indices, 1):
        print(f"\n{i}. Score: {similarities[idx]:.4f} | Category: {y_train.iloc[idx]}")
        print(f"\nText: {X_train.iloc[idx][:120]} ......")
        
sample_msg = "I was charged twice on my credit card for the same order, please refund me."
analyze_new_ticket(sample_msg)
```
